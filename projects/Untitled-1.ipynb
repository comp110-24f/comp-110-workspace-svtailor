```python
import folium
```

```python
m = folium.Map(location=(35.7915, -78.7811), tiles = "Cartodb Positron")
# m is a folium map at the given coordinates and style of the tiles 

folium.Marker(
    location=[35.76330480594747, -78.74700281687525],
    tooltip = "Noodle Blvd",
    popup = "10/10, I love the food", 
    icon = folium.Icon(icon = "heart", color = "darkred")

).add_to(m)
# folium has a lot of built in features that you can use to label and organize your map

m
```
